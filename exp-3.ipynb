```python
import heapq


# ---------- Union-Find for Kruskal ----------
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])   # Path Compression
        return self.parent[x]

    def union(self, x, y):
        root_x = self.find(x)
        root_y = self.find(y)

        if root_x == root_y:
            return False

        if self.rank[root_x] < self.rank[root_y]:
            root_x, root_y = root_y, root_x

        self.parent[root_y] = root_x

        if self.rank[root_x] == self.rank[root_y]:
            self.rank[root_x] += 1

        return True


# ---------- Kruskal Algorithm ----------
def kruskal(n, edges):
    edges.sort()

    uf = UnionFind(n)
    mst = []
    total_cost = 0

    for weight, u, v in edges:
        if uf.union(u, v):
            mst.append((u, v, weight))
            total_cost += weight

            if len(mst) == n - 1:
                break

    return mst, total_cost


# ---------- Prim Algorithm ----------
def prim(n, adj, start=0):
    visited = [False] * n
    parent = [-1] * n
    key = [float("inf")] * n

    key[start] = 0
    pq = [(0, start)]

    mst = []
    total_cost = 0

    while pq:
        weight, u = heapq.heappop(pq)

        if visited[u]:
            continue

        visited[u] = True

        if parent[u] != -1:
            mst.append((parent[u], u, weight))
            total_cost += weight

        for v, wt in adj[u]:
            if not visited[v] and wt < key[v]:
                key[v] = wt
                parent[v] = u
                heapq.heappush(pq, (wt, v))

    return mst, total_cost


# ---------- Main Program ----------
if __name__ == "__main__":

    n = 7

    edges = [
        (7, 0, 1),
        (1, 0, 3),
        (4, 1, 2),
        (9, 1, 3),
        (8, 1, 4),
        (5, 2, 4),
        (10, 3, 4),
        (6, 3, 5),
        (8, 4, 5),
        (3, 4, 6),
        (11, 5, 6)
    ]

    adj = {}

    for weight, u, v in edges:
        adj.setdefault(u, []).append((v, weight))
        adj.setdefault(v, []).append((u, weight))

    kruskal_mst, kruskal_cost = kruskal(n, edges.copy())
    prim_mst, prim_cost = prim(n, adj)

    print("=== Kruskal's MST ===")
    for u, v, weight in kruskal_mst:
        print(f"Edge ({u} - {v}) Weight: {weight}")
    print("Total MST Cost:", kruskal_cost)

    print("\n=== Prim's MST ===")
    for u, v, weight in prim_mst:
        print(f"Edge ({u} - {v}) Weight: {weight}")
    print("Total MST Cost:", prim_cost)
```
